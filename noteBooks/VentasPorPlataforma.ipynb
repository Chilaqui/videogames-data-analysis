```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Leyendo los datos
df = pd.read_csv('../data/vgsales.csv')
df.head()
```

```python
conteo = df['Platform'].value_counts()
conteo
```

```python
#Filtrado de Datos
conteo = df['Platform'].value_counts()
conteo.head()
conteo.plot(kind='bar', figsize=(10, 5))
plt.title('Plataformas Y sus Juegos')
plt.xlabel('Plataforma')
plt.ylabel('Cantidad de juegos')
plt.show()
```
