```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Leer el archivo CSV
df = pd.read_csv('../data/vgsales.csv')
df.head()
```

```python
filtro10 = df.sort_values(by='Global_Sales', ascending=False).head(10)
filtro10.head(10)
```

```python
filtro10 = df.sort_values(by='Global_Sales', ascending=False).head(10)
plt.figure(figsize=(12, 6))
sns.barplot(data=filtro10, x='Name', y='Global_Sales', palette='viridis')
plt.title('Top 10 Juegos Más Vendidos Globalmente')
plt.xlabel('Nombre del Juego')
plt.ylabel('Ventas Globales (millones)')
plt.xticks(rotation=45)
plt.show()
```
